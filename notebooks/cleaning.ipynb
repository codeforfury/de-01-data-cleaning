```python
import pandas as pd

# Load raw Airbnb dataset for initial inspection
df = pd.read_csv("../data/raw/AB_NYC_2019.csv")
df.head()
```

```python
df.info()
```

```python
# Check exact count of missing values per column
df.isnull().sum()
```

```python
# Verify: are last_review nulls always paired with 0 reviews?
df[df['last_review'].isnull()]['number_of_reviews'].unique()
```

```python
# Fill reviews_per_month nulls with 0 — these listings have zero reviews, not missing data
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
```

```python
# Convert last_review to datetime; nulls remain (no review date exists for these listings)
df['last_review'] = pd.to_datetime(df['last_review'])
```

```python
df[['last_review', 'reviews_per_month']].isnull().sum()
```

```python
# Investigate: look at rows where name or host_name is missing
df[df['name'].isnull() | df['host_name'].isnull()]
```

```python
# Fill missing name and host_name with placeholder — negligible % of rows, dropping would lose valid data unnecessarily
df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')
```

```python
df[['name', 'host_name']].isnull().sum()
```

```python
# Check for exact duplicate rows
df.duplicated().sum()
```

```python
# Check how many listings have price = 0
(df['price'] == 0).sum()
```

```python
# Look at the actual rows where price = 0
df[df['price'] == 0]
```

```python
# Drop price=0 rows — these are data entry errors (Airbnb doesn't offer free listings), 
# and imputing a fake price isn't reliable
df = df[df['price'] != 0]
```

```python
(df['price'] == 0).sum()
```

```python
# Check the distribution of minimum_nights to spot extreme values
df['minimum_nights'].describe()
```

```python
# Check how many listings have unrealistically high minimum_nights (e.g., over 365 days)
(df['minimum_nights'] > 365).sum()
```

```python
# Look at the relevant columns for these extreme minimum_nights listings
df[df['minimum_nights'] > 365][['neighbourhood', 'room_type', 'price', 'minimum_nights', 'number_of_reviews', 'availability_365']]
```

```python
# Drop rows with unrealistic minimum_nights (>365 days) — likely data entry errors, not genuine rental terms
df = df[df['minimum_nights'] <= 365]
```

```python
(df['minimum_nights'] > 365).sum()
```
